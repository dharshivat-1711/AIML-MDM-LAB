### Objective
Develop classification models using SVM, Decision Tree, KNN and Naive Bayes. Evaluate their performance using Accuracy, Precision, Recall, F1-Score and Log Loss, and compare their suitability for classification tasks.

## 1. Import Required Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    log_loss,
    confusion_matrix,
    classification_report
)

print("Libraries imported successfully.")
```

## 2. Download and Load Dataset from Kaggle

```python
# Kaggle dataset:
# https://www.kaggle.com/datasets/uciml/breast-cancer-wisconsin-data

# Download the Kaggle dataset directly for Google Colab
!wget -q https://www.kaggle.com/api/v1/datasets/download/uciml/breast-cancer-wisconsin-data -O breast_cancer.zip
!unzip -o -q breast_cancer.zip

df = pd.read_csv("data.csv")

print("Dataset Shape:", df.shape)
display(df.head())
```

## 3. Data Preprocessing

```python
# Remove unnecessary columns
df = df.drop(columns=["id", "Unnamed: 32"], errors="ignore")

# Convert diagnosis into numerical form
# M = Malignant = 1
# B = Benign = 0
df["diagnosis"] = df["diagnosis"].map({
    "M": 1,
    "B": 0
})

X = df.drop("diagnosis", axis=1)
y = df["diagnosis"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nClass distribution:")
print(y.value_counts())
```

## 4. Split Dataset into Training and Testing Sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

## 5. Feature Scaling

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed successfully.")
```

## 6. Train SVM Model

```python
svm_model = SVC(
    kernel="rbf",
    probability=True,
    random_state=42
)

svm_model.fit(X_train_scaled, y_train)

svm_pred = svm_model.predict(X_test_scaled)
svm_prob = svm_model.predict_proba(X_test_scaled)

print("SVM model trained successfully.")
```

## 7. Train Decision Tree Model

```python
dt_model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

dt_model.fit(X_train, y_train)

dt_pred = dt_model.predict(X_test)
dt_prob = dt_model.predict_proba(X_test)

print("Decision Tree model trained successfully.")
```

## 8. Train KNN Model

```python
knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train_scaled, y_train)

knn_pred = knn_model.predict(X_test_scaled)
knn_prob = knn_model.predict_proba(X_test_scaled)

print("KNN model trained successfully.")
```

## 9. Train Naive Bayes Model

```python
nb_model = GaussianNB()

nb_model.fit(X_train_scaled, y_train)

nb_pred = nb_model.predict(X_test_scaled)
nb_prob = nb_model.predict_proba(X_test_scaled)

print("Naive Bayes model trained successfully.")
```

## 10. Evaluate SVM

```python
svm_accuracy = accuracy_score(y_test, svm_pred)
svm_precision = precision_score(y_test, svm_pred)
svm_recall = recall_score(y_test, svm_pred)
svm_f1 = f1_score(y_test, svm_pred)
svm_loss = log_loss(y_test, svm_prob)

print("SVM Performance")
print("Accuracy :", svm_accuracy)
print("Precision:", svm_precision)
print("Recall   :", svm_recall)
print("F1-Score :", svm_f1)
print("Log Loss :", svm_loss)
```

## 11. Evaluate Decision Tree

```python
dt_accuracy = accuracy_score(y_test, dt_pred)
dt_precision = precision_score(y_test, dt_pred)
dt_recall = recall_score(y_test, dt_pred)
dt_f1 = f1_score(y_test, dt_pred)
dt_loss = log_loss(y_test, dt_prob)

print("Decision Tree Performance")
print("Accuracy :", dt_accuracy)
print("Precision:", dt_precision)
print("Recall   :", dt_recall)
print("F1-Score :", dt_f1)
print("Log Loss :", dt_loss)
```

## 12. Evaluate KNN

```python
knn_accuracy = accuracy_score(y_test, knn_pred)
knn_precision = precision_score(y_test, knn_pred)
knn_recall = recall_score(y_test, knn_pred)
knn_f1 = f1_score(y_test, knn_pred)
knn_loss = log_loss(y_test, knn_prob)

print("KNN Performance")
print("Accuracy :", knn_accuracy)
print("Precision:", knn_precision)
print("Recall   :", knn_recall)
print("F1-Score :", knn_f1)
print("Log Loss :", knn_loss)
```

## 13. Evaluate Naive Bayes

```python
nb_accuracy = accuracy_score(y_test, nb_pred)
nb_precision = precision_score(y_test, nb_pred)
nb_recall = recall_score(y_test, nb_pred)
nb_f1 = f1_score(y_test, nb_pred)
nb_loss = log_loss(y_test, nb_prob)

print("Naive Bayes Performance")
print("Accuracy :", nb_accuracy)
print("Precision:", nb_precision)
print("Recall   :", nb_recall)
print("F1-Score :", nb_f1)
print("Log Loss :", nb_loss)
```

## 14. Comparative Performance Table

```python
results = pd.DataFrame({
    "Model": ["SVM", "Decision Tree", "KNN", "Naive Bayes"],
    "Accuracy": [svm_accuracy, dt_accuracy, knn_accuracy, nb_accuracy],
    "Precision": [svm_precision, dt_precision, knn_precision, nb_precision],
    "Recall": [svm_recall, dt_recall, knn_recall, nb_recall],
    "F1-Score": [svm_f1, dt_f1, knn_f1, nb_f1],
    "Log Loss": [svm_loss, dt_loss, knn_loss, nb_loss]
})

display(results)
```

## 15. Plot Model Performance Comparison

```python
results.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
].plot(kind="bar", figsize=(10, 6))

plt.title("Classification Model Performance Comparison")
plt.ylabel("Score")
plt.xlabel("Model")
plt.xticks(rotation=0)
plt.legend()
plt.tight_layout()
plt.show()
```

## 16. Generate Confusion Matrices

```python
models = {
    "SVM": svm_pred,
    "Decision Tree": dt_pred,
    "KNN": knn_pred,
    "Naive Bayes": nb_pred
}

for name, prediction in models.items():
    cm = confusion_matrix(y_test, prediction)

    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")

    plt.title(f"{name} Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()
```

## 17. Find Best Performing Models

```python
best_accuracy_model = results.loc[
    results["Accuracy"].idxmax(), "Model"
]

best_f1_model = results.loc[
    results["F1-Score"].idxmax(), "Model"
]

lowest_loss_model = results.loc[
    results["Log Loss"].idxmin(), "Model"
]

print("Best Accuracy Model  :", best_accuracy_model)
print("Best F1-Score Model  :", best_f1_model)
print("Lowest Log Loss Model:", lowest_loss_model)
```

## 18. Classification Reports

```python
print("===== SVM =====")
print(classification_report(y_test, svm_pred))

print("===== Decision Tree =====")
print(classification_report(y_test, dt_pred))

print("===== KNN =====")
print(classification_report(y_test, knn_pred))

print("===== Naive Bayes =====")
print(classification_report(y_test, nb_pred))
```

## Final Observation

The four classification algorithms are compared using Accuracy, Precision, Recall, F1-Score and Log Loss. The model with higher Accuracy, Precision, Recall and F1-Score and lower Log Loss is considered more suitable for the given classification task. The comparative table and graphs provide a clear basis for selecting the most appropriate algorithm.

# 19. Model-Specific Visualizations

The following graphs show the decision boundaries of SVM, Decision Tree and KNN using two selected features. The full-feature models and their metrics above use all available features.

```python
# Prepare two features only for 2D visualization
visual_features = ["radius_mean", "texture_mean"]

X_vis = df[visual_features]
y_vis = df["diagnosis"]

X_vis_train, X_vis_test, y_vis_train, y_vis_test = train_test_split(
    X_vis,
    y_vis,
    test_size=0.20,
    random_state=42,
    stratify=y_vis
)

vis_scaler = StandardScaler()
X_vis_train_scaled = vis_scaler.fit_transform(X_vis_train)
X_vis_test_scaled = vis_scaler.transform(X_vis_test)

print("Visualization features:", visual_features)
```

## 20. SVM Decision Boundary

```python
svm_vis = SVC(kernel="rbf", probability=True, random_state=42)
svm_vis.fit(X_vis_train_scaled, y_vis_train)

def plot_boundary(model, X, y, title, xlabel, ylabel):
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 300),
        np.linspace(y_min, y_max, 300)
    )

    grid = np.c_[xx.ravel(), yy.ravel()]
    Z = model.predict(grid).reshape(xx.shape)

    plt.figure(figsize=(8, 6))
    plt.contourf(xx, yy, Z, alpha=0.25, cmap="coolwarm")
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm",
                edgecolor="k", s=35)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.tight_layout()
    plt.show()

plot_boundary(
    svm_vis,
    X_vis_train_scaled,
    y_vis_train.to_numpy(),
    "SVM Classification Decision Boundary",
    "radius_mean (scaled)",
    "texture_mean (scaled)"
)
```

## 21. Decision Tree Visualization

```python
dt_vis = DecisionTreeClassifier(
    criterion="entropy",
    max_depth=4,
    random_state=42
)

dt_vis.fit(X_vis_train, y_vis_train)

# Decision boundary
plot_boundary(
    dt_vis,
    X_vis_train.to_numpy(),
    y_vis_train.to_numpy(),
    "Decision Tree Classification Map",
    "radius_mean",
    "texture_mean"
)

# Actual tree structure
from sklearn.tree import plot_tree

plt.figure(figsize=(20, 10))
plot_tree(
    dt_vis,
    feature_names=visual_features,
    class_names=["Benign", "Malignant"],
    filled=True,
    rounded=True,
    fontsize=9
)
plt.title("Decision Tree Structure")
plt.tight_layout()
plt.show()
```

## 22. KNN Decision Boundary

```python
knn_vis = KNeighborsClassifier(n_neighbors=5)
knn_vis.fit(X_vis_train_scaled, y_vis_train)

plot_boundary(
    knn_vis,
    X_vis_train_scaled,
    y_vis_train.to_numpy(),
    "KNN Classification Decision Boundary",
    "radius_mean (scaled)",
    "texture_mean (scaled)"
)
```

# 23. Naive Bayes – Three Required Types

Only the first three common Naive Bayes types are included:

1. Gaussian Naive Bayes
2. Multinomial Naive Bayes
3. Bernoulli Naive Bayes

Each type gets its own performance metrics, confusion matrix and decision-boundary visualization.

```python
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
from sklearn.preprocessing import MinMaxScaler

# Prepare data for the three Naive Bayes variants

# Gaussian: continuous scaled features
X_gaussian_train = X_train_scaled
X_gaussian_test = X_test_scaled

# Multinomial: non-negative features
minmax_scaler = MinMaxScaler()
X_multi_train = minmax_scaler.fit_transform(X_train)
X_multi_test = minmax_scaler.transform(X_test)

# Bernoulli: binary features
train_medians = X_train.median()
X_bern_train = (X_train > train_medians).astype(int)
X_bern_test = (X_test > train_medians).astype(int)

print("Data prepared for Gaussian, Multinomial and Bernoulli Naive Bayes.")
```

## 24. Train and Evaluate the Three Naive Bayes Types

```python
nb_models = {
    "Gaussian NB": (
        GaussianNB(),
        X_gaussian_train,
        X_gaussian_test
    ),
    "Multinomial NB": (
        MultinomialNB(),
        X_multi_train,
        X_multi_test
    ),
    "Bernoulli NB": (
        BernoulliNB(),
        X_bern_train,
        X_bern_test
    )
}

nb_results = []
nb_predictions = {}
nb_probabilities = {}

for name, (model, Xtr, Xte) in nb_models.items():
    model.fit(Xtr, y_train)

    pred = model.predict(Xte)
    prob = model.predict_proba(Xte)

    nb_predictions[name] = pred
    nb_probabilities[name] = prob

    nb_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred),
        "Recall": recall_score(y_test, pred),
        "F1-Score": f1_score(y_test, pred),
        "Log Loss": log_loss(y_test, prob)
    })

nb_results = pd.DataFrame(nb_results)

print("Naive Bayes Types Performance")
display(nb_results)
```

## 25. Confusion Matrices for Three Naive Bayes Types

```python
for name, prediction in nb_predictions.items():

    cm = confusion_matrix(y_test, prediction)

    plt.figure(figsize=(5, 4))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues"
    )

    plt.title(f"{name} Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()
    plt.show()
```

## 26. Naive Bayes Types Performance Comparison

```python
nb_results.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Naive Bayes Types Performance Comparison")
plt.ylabel("Score")
plt.xlabel("Model")
plt.xticks(rotation=0)
plt.legend()
plt.tight_layout()
plt.show()
```

## 27. Decision Boundaries for All Three Naive Bayes Types

```python
# Two-feature data for visualizing the three Naive Bayes decision boundaries
X2 = df[visual_features]
y2 = df["diagnosis"]

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2,
    y2,
    test_size=0.20,
    random_state=42,
    stratify=y2
)

# Gaussian transformation
scaler2 = StandardScaler()
X2_gaussian = scaler2.fit_transform(X2_train)

# Multinomial transformation
minmax2 = MinMaxScaler()
X2_multinomial = minmax2.fit_transform(X2_train)

# Bernoulli transformation
median2 = X2_train.median()
X2_bernoulli = (X2_train > median2).astype(int)

visual_nb_models = [
    ("Gaussian NB", GaussianNB(), X2_gaussian, "gaussian"),
    ("Multinomial NB", MultinomialNB(), X2_multinomial, "multinomial"),
    ("Bernoulli NB", BernoulliNB(), X2_bernoulli, "bernoulli")
]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, (name, model, Xtrain2, transform_type) in zip(
    axes, visual_nb_models
):
    model.fit(Xtrain2, y2_train)

    x_min = X2_train.iloc[:, 0].min() - 1
    x_max = X2_train.iloc[:, 0].max() + 1
    y_min = X2_train.iloc[:, 1].min() - 1
    y_max = X2_train.iloc[:, 1].max() + 1

    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 250),
        np.linspace(y_min, y_max, 250)
    )

    grid_original = pd.DataFrame({
        visual_features[0]: xx.ravel(),
        visual_features[1]: yy.ravel()
    })

    if transform_type == "gaussian":
        grid = scaler2.transform(grid_original)
    elif transform_type == "multinomial":
        grid = minmax2.transform(grid_original)
    else:
        grid = (grid_original > median2).astype(int).to_numpy()

    Z = model.predict(grid).reshape(xx.shape)

    ax.contourf(
        xx, yy, Z,
        alpha=0.25,
        cmap="coolwarm"
    )

    ax.scatter(
        X2_train.iloc[:, 0],
        X2_train.iloc[:, 1],
        c=y2_train,
        cmap="coolwarm",
        edgecolor="k",
        s=25
    )

    ax.set_title(f"{name} Decision Boundary")
    ax.set_xlabel(visual_features[0])
    ax.set_ylabel(visual_features[1])

plt.suptitle("Three Naive Bayes Types – Decision Boundaries", fontsize=15)
plt.tight_layout()
plt.show()
```
