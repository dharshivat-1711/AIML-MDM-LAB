## 1. Theory / Concept

Regression is a supervised machine learning technique used to predict a continuous numerical output from one or more input features.

In this practical:
- **Linear Regression** is used as a baseline regression model.
- **Ridge Regression** uses L2 regularization to reduce the effect of large coefficients.
- **Polynomial Regression** creates polynomial features to model nonlinear relationships.
- **Gradient Descent** is implemented from scratch to optimize the parameters of Linear Regression.

### Performance Metrics
- **MAE:** Mean Absolute Error; lower values indicate smaller average errors.
- **MSE:** Mean Squared Error; penalizes larger errors more strongly.
- **RMSE:** Square root of MSE; expressed in the same unit as the target.
- **R²:** Measures the proportion of target variation explained by the model; higher values generally indicate better fit.

### Gradient Descent
Gradient Descent starts with initial weights and bias. At every iteration, it calculates the prediction error and the gradient of the loss function, then updates the parameters in the opposite direction of the gradient.

## 2. Import Required Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully.")
```

## 3. Load Real-World Dataset

The **Diabetes dataset** supplied by scikit-learn is used as the real-world regression application. It contains 442 samples, 10 input features, and a continuous target variable.

```python
data = load_diabetes()

X = data.data
y = data.target

print("Number of samples:", X.shape[0])
print("Number of features:", X.shape[1])
print("Target shape:", y.shape)
print("\nFeature names:")
print(data.feature_names)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

## 4. Define Regression Models

```python
models = {
    "Linear Regression": LinearRegression(),

    "Ridge Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", Ridge(alpha=1.0))
    ]),

    "Polynomial Regression (Degree 2)": Pipeline([
        ("poly", PolynomialFeatures(
            degree=2,
            include_bias=False
        )),
        ("scaler", StandardScaler()),
        ("model", LinearRegression())
    ])
}

print("Models defined:")
for name in models:
    print("-", name)
```

## 5. Train and Evaluate Regression Models

Each model is trained on the training set and evaluated on the unseen test set using MAE, MSE, RMSE and R².

```python
results = []
predictions = {}

for name, model in models.items():

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    predictions[name] = y_pred

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    results.append([
        name,
        mae,
        mse,
        rmse,
        r2
    ])

results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "MAE",
        "MSE",
        "RMSE",
        "R2 Score"
    ]
)

results_df
```

## 6. Implement Gradient Descent for Linear Regression

The following class implements Linear Regression without using a built-in optimization routine. The Mean Squared Error gradient is used to update the weights and bias.

```python
class GradientDescentLinearRegression:

    def __init__(self, learning_rate=0.03, epochs=5000):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None
        self.loss_history = []

    def fit(self, X, y):

        n_samples, n_features = X.shape

        self.weights = np.zeros(n_features)
        self.bias = 0.0

        for epoch in range(self.epochs):

            predictions = X @ self.weights + self.bias
            error = predictions - y

            # Gradient of Mean Squared Error
            dw = (2 / n_samples) * (X.T @ error)
            db = (2 / n_samples) * np.sum(error)

            # Parameter update
            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

            # Store loss
            loss = np.mean(error ** 2)
            self.loss_history.append(loss)

        return self

    def predict(self, X):
        return X @ self.weights + self.bias

print("Gradient Descent Linear Regression class created.")
```

## 7. Standardize Data and Train Gradient Descent

```python
scaler_gd = StandardScaler()

X_train_scaled = scaler_gd.fit_transform(X_train)
X_test_scaled = scaler_gd.transform(X_test)

gd_model = GradientDescentLinearRegression(
    learning_rate=0.03,
    epochs=5000
)

gd_model.fit(
    X_train_scaled,
    y_train
)

gd_predictions = gd_model.predict(
    X_test_scaled
)

print("Gradient Descent training completed.")
print("Epochs:", gd_model.epochs)
print("Learning rate:", gd_model.learning_rate)
```

## 8. Evaluate Gradient Descent Model

```python
gd_mae = mean_absolute_error(
    y_test,
    gd_predictions
)

gd_mse = mean_squared_error(
    y_test,
    gd_predictions
)

gd_rmse = np.sqrt(gd_mse)

gd_r2 = r2_score(
    y_test,
    gd_predictions
)

print("Gradient Descent Results")
print("------------------------")
print("MAE  :", gd_mae)
print("MSE  :", gd_mse)
print("RMSE :", gd_rmse)
print("R2   :", gd_r2)
```

## 9. Analyze Gradient Descent Convergence

The training loss is plotted against the number of epochs. A decreasing MSE curve indicates that Gradient Descent is reducing the objective function during optimization.

```python
plt.figure(figsize=(9, 5))

plt.plot(
    gd_model.loss_history,
    label="Training MSE"
)

plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.title("Gradient Descent Convergence")
plt.legend()
plt.grid(True)
plt.show()

print("Initial MSE:", gd_model.loss_history[0])
print("Final MSE:", gd_model.loss_history[-1])

if gd_model.loss_history[-1] < gd_model.loss_history[0]:
    print("Observation: Training MSE decreased during optimization.")
else:
    print("Observation: Training MSE did not decrease.")
```

## 10. Final Performance Comparison

```python
gd_results = pd.DataFrame([{
    "Model": "Gradient Descent Linear Regression",
    "MAE": gd_mae,
    "MSE": gd_mse,
    "RMSE": gd_rmse,
    "R2 Score": gd_r2
}])

final_results = pd.concat(
    [results_df, gd_results],
    ignore_index=True
)

final_results.round(4)
```
